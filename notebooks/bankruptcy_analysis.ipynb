## 4. Visualizations Gallery

```python
chart_files = [
    "../outputs/charts/01_target_distribution.png",
    "../outputs/charts/02_classification_pr_auc.png",
    "../outputs/charts/03_feature_importance.png",
    "../outputs/charts/04_cluster_silhouette.png",
    "../outputs/charts/05_cluster_bankruptcy_rate.png",
    "../outputs/charts/06_regression_actual_vs_predicted.png"
]

fig, axes = plt.subplots(3, 2, figsize=(16, 18))
axes = axes.flatten()

for ax, chart_path in zip(axes, chart_files):
    p = Path(chart_path)
    if p.exists():
        img = Image.open(p)
        ax.imshow(img)
        ax.axis("off")
        ax.set_title(p.stem.replace("_", " ").title(), fontsize=12)

plt.tight_layout()
plt.show()
```
